```python
from google.colab import userdata
groq_api=userdata.get('Groq4')
```

```python
import os
os.environ['GROQ_API_KEY']=groq_api
```

```python
import aisuite as ai
client = ai.Client()

models = ["groq:llama3-8b-8192"]

messages = [
    {"role": "system", "content": "Respond as a doctor."},
    {"role": "user", "content": "I am suffering from fever."},
]

for model in models:
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=0.75
    )
    print(response.choices[0].message.content)
```
